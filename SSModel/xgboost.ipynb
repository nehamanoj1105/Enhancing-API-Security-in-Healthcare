```python
import os
import pandas as pd
from xgboost import XGBClassifier
from sklearn.preprocessing import StandardScaler
import joblib

# 🔧 Get full absolute path
base_path = os.path.abspath("SSModel")
os.makedirs(base_path, exist_ok=True)  # Ensure directory exists

# Load your preprocessed dataset
df = pd.read_csv("combined_test.csv")

# Split into features and labels
X_train = df.drop(columns=['label'])
y_train = df['label']

# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_train)

# Train the model
model = XGBClassifier(n_estimators=50, max_depth=3, use_label_encoder=False, eval_metric='logloss')
model.fit(X_scaled, y_train)

# ✅ Save to absolute paths
model_path = os.path.join(base_path, 'xgb_model.pkl')
scaler_path = os.path.join(base_path, 'scaler.pkl')

joblib.dump(model, model_path)
joblib.dump(scaler, scaler_path)

print(f"✅ Saved model to: {model_path}")
print(f"✅ Saved scaler to: {scaler_path}")
```
